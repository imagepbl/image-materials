```python
import warnings

import prism
from prism import Q_
from pathlib import Path

from imagematerials.vehicles.timermodel import TIMERMaterials
```

```python
# Define the complete timeline, including historic tail
time_start = Q_(1960, 'year')
time_end = Q_(2060, 'year')
simulation_start = Q_(1971, 'year')
step = Q_(1, 'year')
complete_timeline = prism.Timeline(time_start, time_end, step)
simulation_timeline = prism.Timeline(simulation_start, time_end, step)

path_scen_data = Path("../data/raw/image/SSP2_baseline")
```

```python
timer_vehicle_model = TIMERMaterials(complete_timeline, path_scen_data) # TODO: fix signature
# 
timer_vehicle_model.simulate(simulation_timeline)
```

```python
# Verify dimensions are set
print(f"Regions: {len(timer_vehicle_model.Region)} - {timer_vehicle_model.Region[:3]}...")
print(f"Types: {len(timer_vehicle_model.Type)} - {timer_vehicle_model.Type[:3]}...")
print(f"Time: {len(timer_vehicle_model.Time)} time steps")
```
